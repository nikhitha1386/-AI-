```python

from heapq import heappush, heappop

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

def heuristic(state):
    return sum(state[i] != goal[i] and state[i] != 0
               for i in range(9))

def solve(start):
    pq = []
    heappush(pq, (heuristic(start), 0, start, []))
    visited = set()

    while pq:
        f, g, state, path = heappop(pq)

        if state == goal:
            for s in path + [state]:
                print(s[:3])
                print(s[3:6])
                print(s[6:])
                print()
            return

        if state in visited:
            continue
        visited.add(state)

        z = state.index(0)
        r, c = divmod(z, 3)

        for dr, dc in [(0, 1), (0, -1),
                       (1, 0), (-1, 0)]:
            nr, nc = r + dr, c + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                nz = nr * 3 + nc
                new = list(state)
                new[z], new[nz] = new[nz], new[z]
                new = tuple(new)

                if new not in visited:
                    heappush(pq, (
                        g + 1 + heuristic(new),
                        g + 1, new, path + [state]
                    ))

    print("No solution found")

solve(start)
```
